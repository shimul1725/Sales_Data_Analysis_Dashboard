```python
from src.data_loader import load_csv, data_audit, preview_dataframe
import pandas as pd

# ---------- Dataset location ----------
FILENAME = 'Superstore.csv'  # <-- change if you use a different name

try:
    df = load_csv(FILENAME)
    rows, cols, miss, dup = data_audit(df)
    print(f'✅ Loaded {FILENAME} – shape: {rows} rows × {cols} cols')
    print('\nMissing values per column:')
    print(miss)
    print(f'\nDuplicate rows: {dup}')
    print('\nFirst 5 rows:')
    display(preview_dataframe(df, 5))
except FileNotFoundError as e:
    print(e)
    print('🛑 Please place the CSV file in data/raw/ and rerun this notebook.')
```
